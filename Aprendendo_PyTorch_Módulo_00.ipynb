## Tipos de dados de tensores

Existem muitos tipos diferentes de [tipos de dados para tensores ](https://docs.pytorch.org/docs/2.15/tensors.html#data-types), alguns são específicos para GPU e outros para CPU. Aprender sobre cada um deles pode levar muito tempo. Normalmente, se virmos `torch.cuda` em algum lugar, o tensor está sendo usado pela GPU (desde que a GPU da NVIDIA usa uma ferramenta chamada CUDA).

O tipo de dado mais comum (e o que é usado como padrão) é o `torch.float32` ou `torch.float`. Isso é refêrenciado como "formato de ponto flutuante de 32 bits".

Mas nós também temos o ponto flutuante de 16 bits (`torch.float16` ou `torch.half`) e o ponto flutuante de 64 bits(`torch.float64` ou `torch.double`).

E para confundir as coisas um pouco mais, temos também os inteiros de 8-bit, 16-bit, 32-bit e 64-bit.

A razão para tudo isso, é a **precisão na computação**. Precisão e a quantia de detalhes usada para descrever um número. Quanto maior é o valor da precisão, mais detalhes e consequentemente mais dados são usados para representar aquele número.

Isso é muito importante em deep learning e computação numérica porque você faz muitas operações: quanto mais dados em alta precisão você precisar calcular, mais pesado fica para a máquina.

Por isso, tipos de dados de menor precisão (como 16-bit) são mais rápidos e leves de computar, otimizando o uso de memória e velocidade da GPU sem sacrificar a acurácia ou a performance do modelo na grande maioria dos casos.

Vamos ver como criar alguns tensores com tipos de dados específicos:

```python
#Tipo de dado padrão para os tensores é float32

float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None, #O padrão é "None", seja torch.float32 ou qualquer outro tipo de dado que seja passado.
                               device=None, #O padrão também é "None", o que útiliza o tipo padrão de tensor, que seria na CPU ao invés da CUDA}
                               requires_grad=False) # Significa que se verdadeiro, as operações feitas nesse tensor serão gravadas

float_32_tensor.shape, float_32_tensor.dtype, float_32_tensor.device
```

Junto do problema de formatos incompativeis, dois dos problemas que mais vamos encarar com o PyTorch são problemas de tipo de dados e dispositivos.

Por exemplo, um dos tensores é  `torch.float32` e o outro `torch.float16` (PyTorch normalmente gosta que trabalhe apenas com um dos formatos)

Ou um dos tensores está na CPU e o outro na GPU(PyTorch prefere calculos entre dois tensores no mesmo dispositivo.)

Agora, vamos criar um tensor com `dtype=torch.float16`

```python
float_16_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=torch.float16,
                               device=torch.device('cuda'),
                               requires_grad=False)

float_16_tensor.shape, float_16_tensor.dtype, float_16_tensor.device
```

## Multiplicação de matrizes

Uma das operações mais utilizadas em álgoritmos de machine learning e deep learning é a multiplicação de matrizes.

O PyTorch implementou a multiplicação de matrizes no método `torch.matmul()`.

As duas regras principais para a mutiplicação de matrizes para se lembrar são :

1. As dimensões internas precisam combinar :     
- `(3,2) @ (3,2)` não vai funcionar
- `(2,3) @ (3,2)` vai funcionar
- `(3,2) @ (2,3)` vai funcionar

2. O formato/shape da matriz resultante, vão ser as dimensões externas
- `(2,3) @ (3,2) -> (2,2)`
- `(3,2) @ (2,3) -> (3,3)`
- `(7,4) @ (4,9) -> (7,9)`

*nota: "@" em Python é o símbolo para multiplicação de matrizes*

Vamos criar um tensor, realizar multiplicação elemento à elemento e multiplicações matriciais nele.

```python
tensor = torch.tensor([1, 2, 3])
tensor.shape
```

A diferença entre a multiplicação elemento à elemento e a multiplicação matricial é a adição de valores

Para a nossa variável tensor com os valores [1,2,3]:

|Operação|Cálculo|Código|
|---|---|---|
|Elemento à elemento|[1 * 1, 2 * 2, 3 * 3]|tensor * tensor
|Mutiplicação matricial|[1 * 1 + 2 * 2 + 3 * 3]|tensor.matmul(tensor)

```python
tensor * tensor
```

```python
tensor.matmul(tensor)
```

```python
tensor @ tensor
```

Podemos fazer a mútiplicação na mão, mas não é recomendado, já que o método `torch.matmul()` é mais rápido.

```python
%%time

value = 0
for i in range(len(tensor)):
  value += tensor[i] * tensor[i]
value
```

```python
%%time
tensor.matmul(tensor)
```

# Um dos erros mais comuns em Deep Learning (Erros de formato)

Já que muito de Deep Learning é multiplicação e realizar operações em matrizes e as matrizes tem regras muitos restritas de que shapes e tamanhos podem ser combinadas, um dos erros mais comuns que vai acontecer em deep learning é o shape mismatches.

```python
try:
  tensor_A = torch.tensor([[1,2],
                          [3,4],
                          [5,6]], dtype=torch.float32)

  tensor_B = torch.tensor([[7,10],
                          [8,11],
                          [9,12]], dtype=torch.float32)

  torch.matmul(tensor_A, tensor_B)
except Exception as erro:
  print(f"Erro ignorado: {erro}")
```

Nós podemos fazer a multiplicação de matrizes entre as matrizes tensor_A e tensor_B funcionar fazendo que suas dimensões internas coincidam.

Uma das formas de fazer isso é com a **transposição** (mudando a dimensão de um tensor específico)

Podemos fazer a transposição no PyTorch usando:
- `torch.transpose(input, dim0, dim1)` onde `input` é o tensor que queremos fazer a transposição e dim0 e dim1 são as dimensões a serem substituidas
- `tensor.T` onde `tensor` é o tensor que queremos fazer a transposição

Vamos tentar a última opção :

```python
print(tensor_A)
print(tensor_B)
```

```python
print(tensor_A)
print(tensor_B.T)
```

```python
#A operação funciona quando o tensor_B é transposicionado

print(f"Shapes originais : tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"Shapes transpostos : tensor_A = {tensor_A.shape} (continua), tensor_B = {tensor_B.T.shape}")
print(f"Multiplicando: {tensor_A.shape} * {tensor_B.T.shape} <- dimensões internas combinam")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"\nOutput shape: {output.shape}")
```

As redes neurais são repletas de multiplicações de matrizes e produtos escalares.

O módulo `torch.nn.Linear()` (veremos em ação mais tarde), também conhecido como camada feed-forward ou camada totalmente conectada, implementa uma multiplicação de matrizes entre uma entrada `x` uma matriz de pesos `A`.

$$y = x ⋅ {A^T} + b$$

Onde :    
- `x` é a entrada para a camada (o deep learning são várias camadas como essa `torch.nn.Linear()` e outras uma sobre as outras).
- `A` é a matriz de pesos criada pela camada; ela começa com um números aleátorios que são ajustados a medida que uma rede neural aprende a representar melhor os padrões nos dados.
- `b` é o termo de viés, usado para compensar ligeiramente os pesos e as entradas.
- `y` é o resultado (uma manipulação da entrada na esperança de descobrir padrões nela)



Essa é uma função Linear (você pode ter visto algo como $y = mx+b$ no ensino médioou em algum outro lugar), e pode ser usada para desenhar uma linha reta!

---



$$y = x ⋅ {A^T} + b$$

Explicando de uma forma mais simples :

1. $x$ (A entrada)
- São os dados de entrada(as características, as features)
- Se você tem 2 clientes, e cada um tem 6 informações, o $x$ é a matriz de shape (2,6) que entra na camada.

2. O $A$ e o $A^T$ (A matriz de pesos)
- O $A$ é a matriz de pesos. o $A^T$ é apenas ela transposta
- O $A$ sempre começa com números aleatórios, essa é a "opinião" inicial da rede neural. Conforme o modelo treina, o PyTorch vai ajustando esses números para que eles aprendam a reconhecer os padrões nos dados (por exemplo, dar mais peso para a idade e menos para o salário na hora de decidir se um cliente é bom ou ruim), esses pesos são impossíveis de definirmos de forma clara, quem define os números é o PyTorch, e inclusive, é muito díficil entender o que cada número significa para a rede neural.

3. O produto ($⋅$ / Multiplicador de matrizes)
- É o ato de múltiplicar os dados dos clientes $(x)$ pelos pesos $(A^T)$. É aqui que a rede combina as informações para gerar novos padrões.

4. $b$ (O viés/ $bias$)
- Se a fórmula fosse apenas $y = mx$ (sem o víes $b$), a reta do gráfico obrigatóriamente teria que passar pelo ponto zero ($x = 0, y = 0$). Ou seja, a IA assumiria que uma casa com 0 metros quadrados, custa exatamente R\$0. Mas na vida real não funciona assim. Pode existir um preço base (um terreno, uma taxa fixa, uma localização) que faz com que, mesmo que o tamanho seja pequeno, o preço comece em um patamar mais alto.
- É aí que entra o viés ($b$): ele é um número somado no final que desloca a reta inteira para cima ou para baixo no gráfico, sem alterar a inclinação dela. Ele dá à rede a liberdade de ajustar o "ponto de partida" da decisão.

Vamos tentar com uma camada linear.

Vamos experimentar alterar o valor de `in_features` e `out_features` abaixo e ver o que acontece.

```python
#Já que uma camada linear começa com uma matriz de pesos aleatórios, vamos tentar reproduzir isso
torch.manual_seed(42)

linear = torch.nn.Linear(in_features=3,
                         out_features=6)

x = tensor_A.T
output = linear(x)
print(f"Input shape: {x.shape}\n")
print(f"Output:\n{output}\n\nOutput shape: {output.shape}")
```

1. Acima, criamos a "máquina" no `torch.nn.Linear`. Aqui estamos criando uma camada linear(uma caixinha com vários pesos aleatórios $A$ e vieses $b$) com duas regras:

- `in_features = 2`: avisa a camada que os dados que vão entrar tem 2 colunas
- `out_features = 6`: avisa a camada que quer que ela devolva 6 saídas diferentes pra cada linha

2. Com o `x` estamos entrando com os dados.

Basicamente foi criada uma matriz de pesos com (2,6), multiplicando com cada número da matriz de x

Mas eis como solucionar: podemos alterar o tipo de dado de um tensor utilizando `torch.Tensor.type(dtype=None)` onde o `dtype` é o tipo de dado que você quer usar

Agora vamos criar um segundo tensor igual o anterior, mas mudando seu tipo de dado para `torch.float16`

E podemos fazer algo parecido com um tensor `torch.int8`

## Remodelando, agrupando, comprimindo e descomprimindo

Algumas vezes, você vai precisar remodelar ou mudar as dimensões dos seus tensores sem alterar os valores dentro dele.

Para isso, esses são os métodos mais populares :

|Método|Descrição|
|---|---|
|`torch.reshape(input,shape)`|Remodela o `input` para o `shape` desde que seja compatível. Também póde ser usado `torch.Tensor.reshape()`
|`Tensor.view()`|Retorna uma visualização do tensor original com um formato diferente, mas compartilha os mesmos dados do tensor original|
|`torch.stack(tensors,dim=0)`|Concatena uma sequencia de tensores em uma nova dimensão. Todos os tensores precisam ter o mesmo tamanho|
|`torch.squeeze(input)`|comprime o `input` para retirar todas as dimensões de valor 1.
|`torch.unsqueeze(input,dim)`|Retorna o `input` com a dimensão de valor 1 adicionada ao `dim`.
|`torch.permuts(input, dims)`| Retorna uma visualização do tensor original `input` com suas dimensões permutadas(re-arranjadas) para o `dims`

Por que fazer essas coisas?

Porque os modelos de deep learning (redes neurais) se baseiam na manipulação de tensores. E devido às regras da multiplicação de matrizes, se houver incompatibilidades de formato, ocorrerão erros. Esses métodos ajudam a garantir que os elementos corretos dos seus tensores sejam combinados com os elementos corretos de outros tensores.

Vamos testa-los



```python
x = torch.arange(1., 8.)
x, x.shape
```

```python
#Adicionando uma dimensão extra
x_reshaped = x.reshape(1, 7)
x_reshaped, x_reshaped.shape
```

```python
#Mudando a visualização com o torch.view()
z = x.view(1, 7)
z, z.shape
```

```python
z[:, 0] = 5
z, x
```

Se você quiser empilhar o seu tensor no topo dele mesmo 5 vezes, podemos fazer com o `torch.stack()`

```python
x_stacked = torch.stack([x, x, x, x], dim=0)
x_stacked, x.shape
```

Que tal remover todas as dimensões únicas de um tensor?

Para isso, você pode usar `torch.squeeze()`(eu me lembro disso como comprimir o tensor para que ele tenha apenas dimensões maiores que 1).


```python
print(f"Tensor anterior: {x_reshaped}")
print(f"Shape anterior: {x_reshaped.shape}")

#Removendo as dimensões de valor 1
x_squeezed = x_reshaped.squeeze()
print(f"\nTensor após squeeze: {x_squeezed}")
print(f"Shape após o squeeze: {x_squeezed.shape}")
#
```

E para reverter o `torch.squeeze()` você pode usar o `torch.unsqueeze()` para adicionar uma dimensão de valor 1 em um index específico.

```python
print(f"Previous tensor: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

## adicionando uma dimensão extra com o unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```
